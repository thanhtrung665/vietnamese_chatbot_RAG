```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Chuẩn bị dữ liệu
# Giả sử đây là kết quả ROUGE-1 của bạn. 
# Bạn có thể thay đổi các giá trị này bằng kết quả thực tế của mình.
data = {
    'Metric': ['Precision', 'Recall', 'F1'],
    'Model Finetune': [0.4266, 0.6966, 0.4163], # Các giá trị từ biểu đồ mẫu của bạn
    'Model Gốc': [0.4063, 0.66350, 0.3168]   # Các giá trị từ biểu đồ mẫu của bạn
}

# 2. Tạo DataFrame
df = pd.DataFrame(data)

# Chuyển đổi DataFrame từ wide format sang long format để phù hợp với Seaborn barplot
# Điều này giúp Seaborn dễ dàng nhóm các cột theo 'Metric' và phân biệt theo 'Model Type'
df_melted = df.melt(id_vars='Metric', var_name='Model Type', value_name='Score')

# In DataFrame đã chuyển đổi để xem cấu trúc
print("DataFrame sau khi melt:")
print(df_melted)
print("\n")

# 3. Vẽ biểu đồ bằng Seaborn và Matplotlib
plt.figure(figsize=(10, 6)) # Thiết lập kích thước của biểu đồ

# Sử dụng seaborn.barplot để vẽ biểu đồ cột đôi
# x='Metric': Các cột sẽ được nhóm theo chỉ số (Precision, Recall, F1)
# y='Score': Chiều cao của cột (giá trị của chỉ số)
# hue='Model Type': Phân biệt các cột bằng màu sắc giữa 'Model Finetune' và 'Model Gốc'
# palette: Bảng màu (có thể chọn 'viridis', 'plasma', 'mako', 'rocket', v.v.)
# Tùy chỉnh màu sắc để giống biểu đồ mẫu (xanh dương cho finetune, xanh lá cho gốc)
# Cụ thể, bạn có thể tự định nghĩa palette như sau:
custom_palette = {"Model Finetune": "#6A5ACD", "Model Gốc": "#2E8B57"} # Màu xanh tím và xanh lá sẫm
# Hoặc gần giống với màu trong ảnh:
custom_palette_sample = {"Model Finetune": "#6495ED", "Model Gốc": "#32CD32"} # CornflowerBlue và LimeGreen

ax = sns.barplot(x='Metric', y='Score', hue='Model Type', data=df_melted, palette=custom_palette_sample)

# 4. Tùy chỉnh biểu đồ
plt.title('Biểu đồ so sánh chỉ số ROUGE-1 Score theo mô hình Finetuned và mô hình Gốc', fontsize=16) # Tiêu đề biểu đồ
plt.xlabel('Chỉ số đánh giá', fontsize=12) # Nhãn trục X
plt.ylabel('Score', fontsize=12) # Nhãn trục Y
plt.ylim(0.0, 1) # Đặt giới hạn cho trục Y để dễ nhìn hơn các thay đổi nhỏ

# Thêm giá trị số lên trên mỗi cột
for p in ax.patches:
    ax.annotate(f'{p.get_height():.4f}', # Định dạng giá trị với 3 chữ số thập phân
                (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='center',
                xytext=(0, 9), # Dịch chuyển văn bản lên trên một chút
                textcoords='offset points',
                fontsize=9,
                color='black')

# Hiển thị lưới để dễ đọc giá trị
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Đặt vị trí chú thích
plt.legend(title='Loại Mô hình', loc='upper right', bbox_to_anchor=(1, 1))

# Điều chỉnh layout để tránh nhãn bị chồng chéo
plt.tight_layout()

# Hiển thị biểu đồ
plt.show()

# Có thể lưu biểu đồ nếu muốn
plt.savefig('rouge_1_chart.png', dpi=300, bbox_inches='tight')
```
